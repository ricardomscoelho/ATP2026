### Tpc 2. À semelhança do que foi feito nas aulas, realize as seguintes tarefas:

#### a) Especifique uma função que dada uma string e uma substring não vazia, calcula  o número de vezes em que a substring aparece na string, sem que haja sobreposição de substrings:

````python
def strCount(s, subs):
    partes = s.split(subs)
    return len(partes) - 1

print(strCount("catcowcat", "cat")) # --> 2
print(strCount("catcowcat", "cow")) # --> 1
print(strCount("catcowcat", "dog")) # --> 0
````

#### b) Especifique uma função que recebe uma lista de números inteiros positivos e devolve o menor produto que for possível calcular multiplicando os 3 menores inteiros da lista:

````python
def produtoM3(lista):
    lista.sort()
    return lista[0]*lista[1]*lista[2]

print(produtoM3([12,3,7,10,12,8,9]))
# Resultado esperado: 168 = 3 * 7 * 8
````

#### c) Especifique uma função que dado um número inteiro positivo, repetidamente adiciona os seus dígitos até obter apenas um dígito que é retornado como resultado:

````python
# Input: 38
# Output: 2
# Explicação: 3 + 8 = 11, 1 + 1 = 2.

# Input: 777
# Output: 3
# Explicação: 7 + 7 + 7 = 21, 2 + 1 = 3.

def reduxInt(n):
    while n >= 10 :
        soma = 0
        while n > 0 :
            soma = soma + n%10
            n = n//10
        n = soma
    return n

print(reduxInt(38))
print(reduxInt(777))
````

#### d) Especifique uma função que recebe duas strings, `string1` e `string2`, e devolve o índice da primeira ocorrência de `string2` em `string1`, caso não ocorra nenhuma vez a função deverá retornar `-1`:

````python
# Invocação: indexOf("Hoje está um belo dia de sol!", "belo")
# Resultado: 13

# Invocação: indexOf("Hoje está um belo dia de sol!", "chuva")
# Resultado: -1

def myIndexOf(s1, s2):
    pos = 0
    while pos <= len(s1) - len(s2):
        if s1[pos:pos + len(s2)] == s2:
            return pos
        pos = pos + 1
    return -1

print(myIndexOf("Hoje está um belo dia de sol!", "belo"))
print(myIndexOf("Hoje está um belo dia de sol!", "chuva"))
````

### tpc-3. A Rede Social

Considere que a informação sobre uma rede social está armazenada numa lista de dicionários.

Cada dicionário, correspondente a um _post_ e tem chaves `id`, `conteudo`, `autor`, `dataCriacao` e `comentarios`.
Por sua vez, `comentarios` é uma lista de dicionários com chaves `comentario` e `autor`.

Considere o seguinte exemplo:

``` 
    MyFaceBook = [{
        'id': 'p1', 
        'conteudo': 'A tarefa de avaliação é talvez a mais ingrata das tarefas que um professor
    tem de realizar...', 
        'autor': 'jcr', 
        'dataCriacao': '2023-07-20', 
        'comentarios': [
            {
                'comentario': 'Completamente de acordo...',
                'autor': 'prh'
            },
            {
                'comentario': 'Mas há quem goste...',
                'autor': 'jj'
            }
        ]},
        {
            'id': 'p2',
            ...
        },
        ...
        ]
```

Defina as seguintes funções de manipulação e consulta da rede social:

````python
MyFaceBook = [{'id' : 'p1' , 'conteudo' : 'A tarefa de avaliação é talvez a mais ingrata das tarefas que um professor tem de realizar...', 'autor' : 'jcr', 'dataCriacao' : '2023-07-20' , 'comentarios' : [{'comentario':'Completamente de acordo...', 'autor' : 'jj'}]},{'id' : 'p2' , 'conteudo' : 'bla, bla, bla', 'autor' : 'lll', 'dataCriacao' : '2023-07-29' , 'comentarios' : [{'comentario':'ble, ble, ble', 'autor' : 'cc'}]}]
````

#### a) `quantosPost`, que indica quantos posts estão registados:

````python
def quantosPost(redeSocial):
    return len(redeSocial)

print(quantosPost(MyFaceBook))
````

#### b)  `postsAutor`, que devolve a lista de posts de um determinado autor:

````python
def postsAutor(redeSocial, autor):
    res = []
    for post in redeSocial :
        if post['autor'] == autor :
            res.append(post)
    return res

print(postsAutor(MyFaceBook,'jcr'))
````

#### c) `autores`, que devolve a lista de autores de posts ordenada alfabeticamente:

````python
def autores(redeSocial):
    res = []
    for post in redeSocial :
        if post['autor'] not in res :
            res.append(post['autor'])
    res.sort()
    return res

print(autores(MyFaceBook))
````

#### d) `insPost`, que acrescenta um novo post à rede social a partir dos parâmetros recebidos e devolve a nova rede social. 
    
O campo `id` devrá ser calculado a partir dos já existentes, por exemplo, se a rede tiver posts com id `p1`, `p2` e `p3`, o novo `id` deverá ser `p4`.

````python
def insPost(redeSocial, conteudo, autor, dataCriacao, comentarios):
    maxId = 0
    for post in redeSocial :
        num = int(post['id'][1:])
        if num > maxId :
            maxId = num
    novoId = 'p' + str(maxId + 1)

    novoPost = {
        'id' : novoId,
        'conteudo' : conteudo,
        'autor' : autor,
        'dataCriacao' : dataCriacao,
        'comentarios' : comentarios
    }

    redeSocial.append(novoPost)

    return redeSocial

print(insPost(MyFaceBook, 'Primeiro post', 'pl', '2023-11-17',{'comentario': 'Olá!','autor' : 'le'}))
````

#### e)  `remPost`, que remove um post da rede, correspondente ao `id` recebido.

````python
def remPost(redeSocial, id):
    for post in redeSocial:
        if post['id'] == id:
            redeSocial.remove(post)
    return redeSocial

print(remPost(MyFaceBook,'p2'))
````

#### f) `postsPorAutor`, que devolve uma distribuição de posts por autor (à semelhança do que foi feito nas aulas).

````python
def postsPorAutor(redeSocial):
    distrib = {}
    for post in redeSocial:
        if post['autor'] in distrib :
            distrib[post['autor']] += 1
        else:
            distrib[post['autor']] = 1
    return distrib

print(postsPorAutor(MyFaceBook))
````

#### g) `comentadoPor`, que recebe um autor e devolve a lista de posts comentados por esse autor.

````python
def comentadoPor(redeSocial, autor):
    res = []
    for post in redeSocial :
        for comentario in post['comentarios'] :
            if comentario['autor'] == autor :
                res.append(post)
    return res

print(comentadoPor(MyFaceBook,'jj'))

````
