### TPC5: Aplicação para gestão de alunos

Considere que o modelo do aluno e da turma têm a seguinte estrutura:

`aluno = (nome, id, [notaTPC, notaProj, notaTeste])`

`turma = [aluno]`

* Cria uma aplicação que coloca no monitor o seguinte menu de operações:
    - 1: Criar uma turma;
    - 2: Inserir um aluno na turma;
    - 3: Listar a turma;
    - 4: Consultar um aluno por id;
    - 8: Guardar a turma em ficheiro;
    - 9: Carregar uma turma dum ficheiro;
    - 0: Sair da aplicação
* No fim de executar a operação selecionada, a aplicação deverá colocar novamente o menu e pedir ao utilizador a opção para continuar;
* Utiliza a tua aplicação para criar uma turma com 5 alunos.

```python
def menu():
    print("""\n********MENU********
- 1: Criar uma turma
- 2: Inserir um aluno na turma
- 3: Listar a turma
- 4: Consultar um aluno por id
- 8: Guardar a turma em ficheiro
- 9: Carregar uma turma dum ficheiro
- 0: Sair da aplicação""")
    
opcao = 1
turma= []
while opcao != 0:
    menu()
    opcao= int(input("Selecione uma das opções: "))
    if opcao == 1:
        turma = []
        print("\nA turma foi criada")

    elif opcao == 2:
        def adicionarAlunos (turma, aluno):
            turma.append(aluno)
            return turma
        nome = input("Digite o nome do aluno: ")
        id = input("Numero mecanográfico: ")
        notaTPC = int(input("Notas do TPC: "))
        notaProj = int(input("Nota do projeto: "))
        notaTeste = int(input("Nota do teste: "))
        
        aluno = (nome, id, [notaTPC, notaProj, notaTeste])
        adicionarAlunos(turma, aluno)

    elif opcao == 3:
        print("\nA turma é cosntituida por estes alunos:\n ")
        for alunos in turma:
            print(alunos[0])
        
    elif opcao == 4:
        def consultarAluno(turma, consultar_id):
            registado = False
            for alunos in turma:
                if alunos[1] == consultar_id:
                    print(alunos)
                    registado = True
            if not registado:
                print(f"\nO id «{consultar_id}» não corresponde a nenhum dos id's da turma.")
        consultar_id = input("Insira o numero mecanográfico do aluno que deseja consultar: ")
        consultarAluno(turma, consultar_id)

    elif opcao == 8:
        def turma_ficheriro (turma):
            with open("turmas.txt", "w") as arquivo:

                for aluno in turma:
                    nome = aluno[0]
                    id = aluno[1]
                    notas = aluno[2]

                    notas_em_str = str(notas[0]) + "#" + str(notas[1]) + "#" + str(notas[2])
                    linha = nome + ";" + id + ";" + notas_em_str + "\n"
                    arquivo.write(linha)     
        turma_ficheriro (turma)

    elif opcao == 9:
        def carregar_turma_ficheiro(turma):
            with open("turmas.txt") as arquivo:

                for linha in arquivo:
                    nova_linha = linha.strip()
                    partes = nova_linha.split(";")
                    notas_str = partes[2].split("#")
                    notas_int = [float(n) for n in notas_str]
                    aluno = (partes[0], partes[1], notas_int)
                    turma.append(aluno)
                    return turma
        turma = carregar_turma_ficheiro(turma)
    

print("\nObrigado por utilizar a aplicação")
```
