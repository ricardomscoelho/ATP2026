### Jogo: Corrida para o 100

O total começa em 0. O jogador e o computador alternam somando um número de 1 a 10 ao total. Quem atingir exatamente o número 100 vence.
- Se o computador jogar primeiro deverá vencer;
- Se o computador jogar em segundo poderá vencer se o outro jogador fizer uma aposto fora da estratégia vencedora.

Implementa o jogo com 2 vertentes: o computador joga primeiro (deverá ganhar sempre), e o computador joga em segundo lugar (poderá ganhar ou não dependendo das jogadas do outro).

```python
import random

a = input("Vamos jogar o jogo de corrida para o 100?")
if a==("Sim") or a==("sim") or a==("s"):
    while True:
        print ("""-------------------------------
      Corrida para o 100
-------------------------------
Cada jogador pode adicionar um número de 1 a 10.
O primeiro a chegar a 100 ganha.
1 - Computador joga primeiro
2 - Jogador joga primeiro""")

        opcao = int(input("Escolha uma opção: "))
        total = 0
        if opcao == 1:
            print("O computador joga primeiro!")
            computador = 1
            total = total + computador
            print("Computador adicionou:", computador)
            print("Total:", total)

            while total < 100:
                jogador = int(input("Escolha um número entre 1 e 10: "))
                while jogador < 1 or jogador > 10:
                    print("Jogada inválida!")
                    jogador = int(input("Escolha um número entre 1 e 10: "))
                total = total + jogador
                print("Jogador adicionou:", jogador)
                print("Total:", total)

                computador = 11 - jogador
                total = total + computador
            
                print("Computador adicionou:", computador)
                print("Total:", total)
            
                if total == 100:
                    print("""
**Eu ganhei!!**

O jogo vai reiniciar.""")
                    break
        if opcao == 2:
            print("\nO jogador joga primeiro!")        

            while total < 100:
                jogador = int(input("Escolha um número entre 1 e 10: "))
                while jogador < 1 or jogador > 10 or (total + jogador > 100):
                    print("Jogada inválida, ou o valor excede 100!Escolhe outro valor")
                    jogador = int(input("Escolha um número entre 1 e 10: "))

                total = total + jogador

                print("Jogador adicionou:", jogador)
                print("Total:", total)
                if total == 100:
                    print("""
**Tu ganhaste!!**
                    
O jogo vai reiniciar.""")
                    break
                jpc = (12 - (total % 11)) % 11
                if jpc == 0:
                    computador = 1
                else:
                    computador = jpc
                
                total = total + computador
                    
                print("Computador adicionou:", computador)
                print("Total:", total)            
                if total == 100:
                    print("""
**Eu ganhei!!**
                    
O jogo vai reiniciar.""")
                    break
        else:
            print("Opção inválida, escolha uma das opções indicadas.")
            continue

        c = input("Desesja continuar?")
        if c==("Sim") or c==("sim") or c==("s"):
            continue
        else:
            print("Está bem até à próxima")
            break
else:
    print ("Está bem até à próxima!")
```
