### Jogo "Adivinha o número"

Crie um programa em Python para jogar o jogo "Adivinha o número":
    - O jogo pode ter 2 modalidades: computador pensa num número (entre 0 e 100), utilizador tenta adivinhar; ou, o utilizador pensa num número (entre 0 e 100) e o computador tenta adivinhar;
    - Quem tenta adivinhar responde com uma das afirmações: "Acertou", "O número que pensei é Maior" ou "O número que pensei é Menor";
    - Uma vez descoberto o número o programa deverá terminar imprimindo o número de tentativas que quem adivinhou usou para chegar ao resultado.

```python
import random

a = input("Vamos jogar o jogo de adivinhar um número?")

if a==("Sim") or a==("sim") or a==("s"):
    while True:
        b = input("Preferes que eu escolher um número ou prefers que eu escolha?")

        if b==("Eu adivinho") or b==("Tu escolhes") or b==("eu adivinho") or b==("tu escolhes"):
            print("Ótimo, vou escolher um número entre 0 e 100.")
            n = random.randint(0, 100)
            p = -1
            t = 0
            while p != n:
                t = t + 1
                p = int(input("Qual o teu palpite?"))
                if p > n:
                    print(f"O número é menor que {p}")
                elif p < n:
                    print(f"O número é maior que {p}")
                else:
                    print(f"Acertaste, o número é {n}, precisaste de {t} tentativas!")
        if b==("Eu escolho") or b==("Tu adivinhas") or b==("eu escolho") or b==("tu adivinhas"):
            print ("Ótimo escolhe um número entre 0 e 100 para eu adivinhar, e não me digas  qual é")
            input("Pessiona Enter quando tiveres pronto.")
            limite_inferior = 0
            limite_superior = 100
            r = ""
            t = 0
            while r != "3":
                t = t + 1
                q = (limite_inferior + limite_superior)  // 2
                print (f"O meu palpite é {q}.")
                print ("1 - O número que escolhi é menor.")
                print ("2 - O número que escolhi é maior.")
                print ("3 - Acertei o numero que escolheste.")
                r = input("Escolhe a opção ( 1 | 2 | 3 )")
                if r == "1":
                    limite_inferior = q + 1
                elif r == "2":
                    limite_superior = q -1
                elif r == "3":
                    print(f"Acertei o teu número ({q}) em {t} tentativas")
                else:
                    print ("Opção inválida, escolhe 1, 2 ou 3")

        c = input("Desesja continuar?")
        if c==("Sim") or c==("sim") or c==("s"):
            continue
        else:
            print("Está bem até à próxima")
            break
elif a==("Não") or a==("não") or a==("n") or a==("Nao") or a==("nao"):
    print ("Está bem até à próxima!")

```
